# **Organise the data to test and train folders**

```python
import os, shutil, random


SOURCE = "/content/drive/MyDrive/dataset"

# Target YOLO structure
BASE = "/content/yolo_dataset"
train_img = BASE + "/images/train"
test_img  = BASE + "/images/test"
train_lbl = BASE + "/labels/train"
test_lbl  = BASE + "/labels/test"

# Create directories
for d in [train_img, test_img, train_lbl, test_lbl]:
    os.makedirs(d, exist_ok=True)

all_images = [f for f in os.listdir(SOURCE) if f.endswith(".png")]
random.shuffle(all_images)

# 90% train, 10% test
split_idx = int(len(all_images) * 0.9)
train_files = all_images[:split_idx]
test_files  = all_images[split_idx:]

def copy_pair(filename, dest_img, dest_lbl):
    img_src = os.path.join(SOURCE, filename)
    lbl_src = img_src.replace(".png", ".txt")

    shutil.copy(img_src, dest_img)
    if os.path.exists(lbl_src):
        shutil.copy(lbl_src, dest_lbl)

for f in train_files:
    copy_pair(f, train_img, train_lbl)

for f in test_files:
    copy_pair(f, test_img, test_lbl)

print("Train images:", len(train_files))
print("Test images:", len(test_files))
```

```python
yaml = """
train: /content/drive/MyDrive/dataset/yolo_dataset/images/train
val: /content/drive/MyDrive/dataset/yolo_dataset/images/test

nc: 2
names: ['bottle', 'can']
"""

with open('/content/yolov5/cans_bottles.yaml', 'w') as f:
    f.write(yaml)

print(open('/content/yolov5/cans_bottles.yaml').read())
```
